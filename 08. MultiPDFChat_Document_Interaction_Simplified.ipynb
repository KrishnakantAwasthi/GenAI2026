# **Retrieval-Augmented Generation (RAG)**

### 🧠 **What is RAG?**



**RAG** stands for **Retrieval-Augmented Generation**.

It is a technique where a language model (like ChatGPT or Gemini) first **retrieves relevant information** from documents or databases, and then **generates an answer** using that information.

> In simple terms: RAG = Search + Generate

**Why RAG?**


Traditional language models are limited to what they learned during training. If you want to answer based on **external or private knowledge (like PDFs, docs, wikis) without fine-tuning, RAG is the answer.**

RAG = Retrieval + Generation = Smart, Factual, and Custom LLM Responses

###**Key Components in the RAG Architecture**

#### **Pre-processing (Indexing Phase)**

**🎯 Goal**: Convert raw HR documents into searchable vector representations.

- **User Documents**  
  → HR Handbook, Leave Policy PDF, HR FAQs

- **Embedding Model**  
  → Converts chunks of these documents into vector format

- **Embeddings**  
  → Each chunk becomes a numeric vector that captures its meaning

- **Vector DB**  
  → All embeddings are stored here for fast semantic search

> ✅ This step prepares your HR documents so the system can later search for relevant answers.

#### **Retrieval Phase**

**🎯 Goal**: Find the most relevant document chunks that can answer the user's query.

- **User Input**  
  → _"How many casual leaves do I get?"_

- **Query Embedding**  
  → The system turns this question into a vector using the same embedding model

- **Vector Search**  
  → Compares the question vector with all document vectors in the Vector DB

- **Relevant Context**  
  → Retrieves chunks like:  
  _"Employees are entitled to 12 casual leaves per calendar year..."_

#### **Augmentation Phase**

**🎯 Goal**: Add useful context to the user’s question before passing it to the AI model.

- Combine the **query** + **retrieved context**
- Example Input to LLM:  
  > **Query**: "How many casual leaves do I get?"  
  > **Context**: "As per the leave policy, employees are eligible for 12 casual leaves annually..."


#### **Generation Phase**



**🎯 Goal**: Use the combined input to generate a clear and helpful answer.

- The **LLM (like GPT)** receives the query + context
- It generates the final response:  
  _"You are eligible for 12 casual leaves per year as per company policy."_

- The response is sent back to the user

### **What is a RAG Chain?**



The **RAG Chain** is the **core runtime workflow** in a Retrieval-Augmented Generation (RAG) system.

It activates **after data is indexed**, and is responsible for:
- Retrieving relevant information from the vector database
- Augmenting the user query with retrieved context
- Generating a response using a Large Language Model (LLM)

#### ⚙️ **How the RAG Chain Works (Step-by-Step)**



> Example Query: "How many casual leaves do I get?"

1. **Query Embedding**  
   - The user's question is converted into a vector using the **same embedding model** used during indexing.

2. **Vector Search**  
   - The system searches the **Vector Database** to find similar document chunks.

3. **Retrieve Context**  
   - The top relevant chunks (e.g., from HR policy) are retrieved.

4. **Augmentation**  
   - The system combines:
     - The original query  
     - The retrieved content  
     → into a single prompt

5. **LLM Generation**  
   - The combined input is sent to the **LLM** (e.g., GPT-4, Gemini)
   - It generates a fact-based, context-aware response

6. **Return Answer to User**  
   - _Example output: "You are eligible for 12 casual leaves per year."_

#### **STEP 5: Extract Text from PDFs**

Reads multiple PDFs and returns their combined text content.

````python
def get_pdf_text(pdf_paths):
    text = ""
    for pdf_path in pdf_paths:
        pdf_reader = PdfReader(pdf_path) # Create a PdfReader object for the current PDF file
        for page in pdf_reader.pages: # Iterate through each page in the PDF
            text += page.extract_text() # Extract text from the page and append it to the 'text' variable
    return text # Return the concatenated text from all PDFs
````

#### **STEP 6: Split Text into Chunks**

Splits long text into overlapping chunks to preserve context (important for QA).

````python
def get_text_chunks(text):
    text_splitter = RecursiveCharacterTextSplitter(chunk_size=10000, chunk_overlap=1000)
    # Initialize the RecursiveCharacterTextSplitter with a chunk size of 10,000 characters
    # overlap of 1,000 characters between consecutive chunks to maintain context.
    chunks = text_splitter.split_text(text) # Split the text into chunks
    return chunks # Return the list of text chunks

    # I love apple -> s1
    # Wow, Even i love apple -> s2
````

#### **STEP 8: Create Conversational Chain (QA logic)**

````python
def get_conversational_chain():
    prompt_template = """
    Answer the question as detailed as possible from the provided context, make sure to provide all the details, if the answer is not in
    provided context just say, "answer is not available in the context", don't provide the wrong answer\n\n
    Context:\n {context}?\n
    Question: \n{question}\n
    Answer:
    """
    model = ChatGoogleGenerativeAI(model="gemini-1.5-flash", temperature=0.3)  # Initialize the AI model with specified parameters
    prompt = PromptTemplate(template=prompt_template, input_variables=["context", "question"]) # Create a prompt template

    chain = load_qa_chain(model, chain_type="stuff", prompt=prompt) # Load the QA chain with the model and prompt
    return chain  # Return the configured question-answering chain
````

####**STEP 9: Handle User Questions**

- Retrieves relevant chunks from FAISS using the user's question

- Sends it to Gemini with the prompt template

- Prints the AI-generated answer

````python
# Retrieval: In the user_input function
def user_input(user_question):
    # Retrieval: Perform a similarity search on the vector store
    embeddings = GoogleGenerativeAIEmbeddings(model="models/embedding-001")  # Create embeddings for the search

    new_db = FAISS.load_local("/content/drive/MyDrive/faiss_index", embeddings, allow_dangerous_deserialization=True) # Load the FAISS vector store
    docs = new_db.similarity_search(user_question) # Retrieve the most relevant documents based on the user's question

    # Generation: Generate a response using the conversational chain

    chain = get_conversational_chain() # Get the configured conversational chain

    response = chain(
        {"input_documents": docs, "question": user_question}, # Pass retrieved docs and question to the chain
        return_only_outputs=True) # Return only the output from the chain
    print(response) # Print the full response object for debugging
    print("Reply: ", response["output_text"]) # Print the generated reply text
````

### 🛡️ **Adding Guardrails to Our RAG Application**

Now that our AI can read and chat with our PDFs, we face a new problem: **How do we keep the AI focused?** If a user asks our PDF chatbot for a pancake recipe, the AI might just answer it! In an enterprise setting, you want your bot to stay strictly on-topic and only use the provided documents. To do this, we implement **Guardrails**.

Guardrails are safety checkpoints in our pipeline. In RAG, they typically happen in three places:
1. **Input Guardrails:** Checking the user's prompt *before* we search the PDFs. (e.g., "Is this question safe and relevant?")
2. **Retrieval Guardrails:** Filtering the search results. (e.g., "Only use document chunks that have a 80% or higher relevance match.")
3. **Output Guardrails:** Checking the AI's final answer. (e.g., "Did the AI actually use the PDF context, or did it hallucinate?")

#### 🛑 **Implementing an Input Guardrail**

Before we even search our PDFs or use tokens to generate a long answer, we should check if the user's question is relevant or safe. Here is a simple program logic using a quick, low-temperature LLM call to classify the intent of the user's prompt.

```python
def is_query_relevant(user_query, expected_topic="company policies and HR documents"):
    """
    Acts as an Input Guardrail to block irrelevant questions.
    """
    # Use a fast, cheap model for quick classification
    classifier_model = genai.GenerativeModel('gemini-1.5-flash')
    
    classification_prompt = f"""
    You are a strict guardrail system.
    A user has asked the following question: "{user_query}"
    Does this question logically relate to {expected_topic}?
    
    Reply with ONLY the word "YES" or "NO". Do not include any other text.
    """
    
    # Temperature 0 ensures strict, predictable outputs
    response = classifier_model.generate_content(
        classification_prompt,
        generation_config=genai.types.GenerationConfig(temperature=0.0)
    )
    
    result = response.text.strip().upper()
    return result == "YES"

#### 🔍 **Implementing a Retrieval Guardrail (Similarity Threshold)**

Sometimes a user asks a relevant question, but our vector database simply doesn't have the answer in the PDFs. If we pass low-quality or irrelevant chunks to the LLM, it will try to guess or hallucinate.

We can fix this by enforcing a **Similarity Score Threshold**. We only allow document chunks that score highly on relevance to pass through to the AI.

```python
def retrieve_with_guardrail(vector_database, user_query, min_score_threshold=0.75):
    """
    Acts as a Retrieval Guardrail by filtering out poor context matches.
    (Note: Syntax varies slightly depending on whether you use FAISS, Chroma, etc.)
    """
    # Ask the database for the top 4 most relevant chunks with their scores
    results_with_scores = vector_database.similarity_search_with_relevance_scores(user_query, k=4)
    
    valid_chunks = []
    
    # Filter the results
    for chunk, score in results_with_scores:
        if score >= min_score_threshold:
            valid_chunks.append(chunk.page_content)
            
    if not valid_chunks:
        return None # Prevent the LLM from generating an answer blindly
        
    return "\n\n".join(valid_chunks)

# --- Testing the Retrieval Guardrail ---
context = retrieve_with_guardrail(my_vector_db, "What is the remote work policy?", min_score_threshold=0.80)

if context is None:
    print("Guardrail triggered: I couldn't find enough relevant information in the documents to answer that securely.")
else:
    print("High-quality context found. Passing to LLM...")

#### 🛡️ **Implementing an Output Guardrail (Context Verification)**

Even with great context, an LLM might accidentally slip in outside knowledge. An **Output Guardrail** uses a secondary, automated check to ensure the LLM's final answer is strictly supported by the PDF text.

```python
def verify_output_guardrail(retrieved_context, llm_generated_answer):
    """
    Acts as an Output Guardrail to prevent hallucinations.
    """
    evaluator_model = genai.GenerativeModel('gemini-1.5-flash')
    
    evaluation_prompt = f"""
    You are an auditing AI.
    You must determine if the GENERATED_ANSWER is completely supported by the provided CONTEXT.
    If the answer includes facts or claims not found in the context, it is a failure.
    
    CONTEXT:
    {retrieved_context}
    
    GENERATED_ANSWER:
    {llm_generated_answer}
    
    Reply with ONLY "PASS" if it is fully supported, or "FAIL" if it hallucinates or uses outside knowledge.
    """
    
    response = evaluator_model.generate_content(
        evaluation_prompt,
        generation_config=genai.types.GenerationConfig(temperature=0.0)
    )
    
    result = response.text.strip().upper()
    return result == "PASS"

# --- Testing the Output Guardrail ---
# Imagine our LLM generated this response:
draft_answer = "Employees get 20 vacation days. Also, the CEO loves playing golf."
# But the PDF context only said: "Employees get 20 vacation days."

if verify_output_guardrail(actual_pdf_context, draft_answer):
    print("Bot:", draft_answer)
else:
    print("Guardrail triggered: The generated response could not be verified against the source documents. Retrying or aborting...")

### **Best Practices**

- Always clean and preprocess text data from PDFs before embedding (remove headers, footers, and irrelevant content).
- Use efficient chunking strategies (e.g., overlapping sliding windows) to preserve context in long documents.
- Choose a well-suited embedding model to represent text semantically for accurate retrieval.
- Limit the number of documents or chunk size to ensure fast and relevant retrieval in vector search.
- Implement session-based memory in Streamlit to maintain chat context between queries.
- Test with real user queries to ensure your RAG system retrieves and generates useful, fact-based answers.
- Log errors and fallback to a generic response if no relevant context is retrieved.
- Keep the UI clean and intuitive for non-technical users to interact with uploaded documents.


### ❌ **Shortcomings**


- The chatbot's accuracy heavily depends on the quality and clarity of the uploaded PDFs.
- Retrieval may fail if the embedded context is too generic or the chunking strategy is not optimal.
- Responses can become hallucinated if the model generates answers beyond the retrieved context.
- Lack of proper session memory may result in disconnected or repetitive answers in multi-turn conversations.
- Performance may degrade with very large document sets due to vector store size and query latency.
- Limited control over the LLM's output can lead to inconsistent tone or verbosity in responses.
- Not all file types or PDF formats (e.g., scanned images or tables) are handled equally well.

## **Mnemonic**

**RAG = Retrieve, Augment, Generate**

- **Retrieve**: Search for relevant document chunks using a retriever from a vector database.
- **Augment**: Combine the retrieved information with the original query to provide rich context.
- **Generate**: Use a language model to generate a final answer based on the augmented input.

This simple mnemonic helps remember the three-step process at the heart of any Retrieval-Augmented Generation system.
